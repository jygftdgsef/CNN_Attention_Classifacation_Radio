```python
import sys
import torch
import torchvision
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Python utilisé :", sys.executable)
print("Version de Python :", sys.version.split()[0])
print("PyTorch :", torch.__version__)
print("Torchvision :", torchvision.__version__)

# Nous utiliserons le processeur pour les calculs.
device = torch.device("cpu")
print("Matériel de calcul :", device)

# Vérification d'un calcul simple.
valeurs = torch.tensor([1, 2, 3], device=device)
print("Résultat du calcul :", valeurs * 2)

# Vérification de l'affichage des graphiques.
plt.plot([1, 2, 3], [2, 4, 6], marker="o")
plt.title("Vérification de l'affichage")
plt.xlabel("Valeur initiale")
plt.ylabel("Valeur multipliée par 2")
plt.show()
```
